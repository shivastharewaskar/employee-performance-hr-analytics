```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)
print('Libraries successfully imported!')
```

### 1. Load the Dataset

```python
df_raw = pd.read_csv('../01_Dataset/employee_data_raw.csv')
print(f'Raw dataset dimensions: {df_raw.shape}')
df_raw.head()
```

### 2. Dataset Exploration: Columns, Data Types & Info

```python
df_raw.info()
```

### 3. Handle Missing Values and Duplicates

```python
print('Missing values before cleaning:')
print(df_raw.isnull().sum())

# Deduplication
df_clean = df_raw.drop_duplicates(subset=['EmployeeID']).copy()

# Imputation
df_clean['Salary'] = df_clean.groupby('Department')['Salary'].transform(lambda x: x.fillna(x.median()))
df_clean['WorkMode'] = df_clean['WorkMode'].fillna(df_clean['WorkMode'].mode()[0])

print('\nRemaining missing values after cleaning:')
print(df_clean.isnull().sum())
print(f'Final Cleaned Shape: {df_clean.shape}')
```

### 4. Descriptive Statistical Analysis

```python
df_clean.describe().round(2)
```

### 5. Department-Wise & Salary Analysis

```python
dept_summary = df_clean.groupby('Department').agg({
    'EmployeeID': 'count',
    'Salary': ['mean', 'median', 'max'],
    'YearsOfExperience': 'mean',
    'Attrition': lambda x: (x == 'Yes').mean() * 100
}).round(2)
dept_summary
```

### 6. Correlation Analysis

```python
num_cols = ['Age', 'YearsOfExperience', 'YearsAtCompany', 'Salary', 'PerformanceRating', 'JobSatisfaction', 'WorkLifeBalance']
plt.figure(figsize=(8, 6))
sns.heatmap(df_clean[num_cols].corr(), annot=True, fmt='.2f', cmap='Blues')
plt.title('Key Workforce Metrics Correlation')
plt.show()
```
